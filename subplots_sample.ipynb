```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(18, 6))
axs = axes.flatten()

X = np.arange(0,24)

for n in np.arange(0,8):
    
    Y1 = np.random.rand(24)
    Y2 = np.random.rand(24) 
    axs[n].plot(X, Y1, 'k', label = 'line 1')
    axs[n].plot(X, Y2, 'r', label = 'line 2')

    axs[n].set_title('Plot number '+str(n+1))

    axs[n].set_ylabel('Time')
    axs[n].set_xlabel('Values')

handles, labels = axs[0].get_legend_handles_labels()
unique = dict(zip(labels, handles))
fig.subplots_adjust(right=0.9)
fig.legend(unique.values(), unique.keys(), loc="right", fontsize=16)

fig.subplots_adjust(hspace=0.4)
fig.subplots_adjust(wspace=0.4)
```
